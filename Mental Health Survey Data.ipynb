## **KPI's REQUIREMENT**

```python
# Treatment Rate KPI
treatment_rate=df['Treatment'].value_counts(normalize=True) * 100

# Stress Level Distribution KPI
stress_distribution=df['IncreasingStress'].value_counts(normalize=True) * 100

# Mood Swings KPI
mood_swings=df['MoodSwings'].value_counts(normalize=True) * 100

# Coping Struggles KPI
coping_struggles=df['CopingStruggles'].value_counts(normalize=True) * 100

# Social Weakness KPI
social_weakness=df['SocialWeakness'].value_counts(normalize=True) * 100

# Work Interest KPI
Work_Interest=df['WorkInterest'].value_counts(normalize=True) * 100

#Display
print("Treatment Rate:",treatment_rate)
print("\nStress Level Distribution (%):",stress_distribution)
print("\nMood Swings (%):",mood_swings)
print("\nCoping Struggles (%):",coping_struggles)
print("\nSocial Weakness (%):",social_weakness)
print("\nWork Interest (%):",Work_Interest)
```

## **CHARTS REQUIREMENT**

#### **🔹 Gender Distribution Analysis**

```python
# Create figure with custom size (width=6, height=4)
plt.figure(figsize=(6,4))

# Set background style to white grid for better readability
sns.set_style("whitegrid")

# Create count plot for Gender column with colorful 'Set2' palette
sns.countplot(x='Gender', data=df, palette='Set2')

# Add title with larger font and bold style
plt.title('Gender Distribution', fontsize=14, fontweight='bold')

# Label X and Y axes
plt.xlabel('Gender')
plt.ylabel('Count')

# Display the plot
plt.show()
```

#### insights
The dataset has more male respondents compared to female respondents, indicating unequal participation in the survey.

Since male representation is higher, overall mental health trends observed in the dataset may reflect male responses more strongly.

Higher male count does not necessarily mean males experience more mental health issues — it only shows survey participation distribution.


#### **🔹 Country-wise Participation**


```python

# Create a figure with width=10 and height=5
plt.figure(figsize=(10,4))

# Get top 10 countries based on number of participants
top_countries = df['Country'].value_counts().head(10)

# Create a line plot with markers
plt.plot(top_countries.index,
         top_countries.values,
         marker='o')

# Rotate country names for better readability
plt.xticks(rotation=45)

# Label the axes
plt.xlabel('Country')
plt.ylabel('Number of Participants')

# Add title Display the plot
plt.title('Top 10 Participating Countries')
plt.grid(True)
plt.show()
```

#### insights
The USA dominates the dataset. This means most responses come from the United States.
There is a big gap between USA and UK, and an even larger gap after UK. This shows uneven global participation.

#### **🔹Days Indoor vs Stress**

```python
# Create figure with specific size
plt.figure(figsize=(6,4))

# Set background style
sns.set_style("whitegrid")

# Count plot
sns.countplot(x='DaysIndoors',
              hue='IncreasingStress',
              data=df,
              palette='Set2')

# Add bold title
plt.title('Increasing Stress vs Days Indoors', fontsize=14, fontweight='bold')

# Label axes
plt.xlabel('Days Indoors')
plt.ylabel('Count')

# Add legend
plt.legend(title='Increasing Stress')

# Rotate x labels if needed
plt.xticks(rotation=45)

# Display plot
plt.show()
```

#### insights
* The "Yes" (Increasing Stress) count is highest for:1–14 days,31–60 days

  This suggests that extended but not extreme isolation may increase stress levels.

* “Maybe” is Very High for Long-Term Indoor Stay For More than 2 months, the "Maybe" category is very high.

   This indicates uncertainty or fluctuating stress levels among people indoors long-term.

* Going Out Every Day Still Shows High Stress

  Surprisingly, "Go out Every day" also shows high stress (especially "Maybe").

#### **🔹 Stress Level Analysis**

```python
plt.figure(figsize=(6,4))

sns.set_style("whitegrid")
ax = sns.countplot(x='IncreasingStress',
                   data=df,
                   palette='Set3')

# Add value labels on bars
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}',
                (p.get_x() + p.get_width()/2., p.get_height()),
                ha='center', va='bottom')

plt.title('Stress Level Analysis', fontsize=14, fontweight='bold')
plt.xlabel('Increasing Stress')
plt.ylabel('Count')

plt.show()
```

#### insights
The stress level analysis provides a clear view of how many individuals are facing increasing stress.

The visualization shows the distribution clearly with value labels,

making it easier to interpret the severity of stress within the dataset.

#### **🔹Gender vs Treatment**

```python
# Create a figure with specified size
plt.figure(figsize=(6,4))

# Set background style with grid for better readability
sns.set_style("whitegrid")

sns.countplot(x='Gender',hue='Treatment',data=df,palette='Set2')

# Add title
plt.title('Gender vs Treatment', fontsize=14, fontweight='bold')

# Label axes
plt.xlabel('Gender')
plt.ylabel('Count')

# Add legend with title
plt.legend(title='Treatment')

# Display the plot
plt.show()
```

#### **🔹 Coping Struggles & Mood Swingss**

```python
# Create figure
plt.figure(figsize=(6,4))

# Set style
sns.set_style("whitegrid")

# Count plot comparing Coping Struggles and Mood Swings
sns.countplot(x='CopingStruggles',
              hue='MoodSwings',
              data=df,
              palette='Set2')

# Add title and labels
plt.title('Coping Struggles vs Mood Swings', fontsize=14, fontweight='bold')
plt.xlabel('Coping Struggles')
plt.ylabel('Count')

plt.legend(title='Mood Swings')
plt.show()
```

This highlights the importance of emotional regulation strategies.

Mental health support systems can help improve coping ability.
